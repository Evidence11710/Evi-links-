```python
# Load CIFAR-10 dataset

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("Training images:", X_train.shape)

print("Training labels:", y_train.shape)

print("Testing images:", X_test.shape)

print("Testing labels:", y_test.shape)
```

```python
# Normalize pixel values to between 0 and 1

X_train = X_train.astype('float32') / 255.0

X_test = X_test.astype('float32') / 255.0

print("Training data normalized")

print("X_train shape:", X_train.shape)

print("X_test shape:", X_test.shape)
```

```python
# Build the CNN model

model = tf.keras.Sequential([

    tf.keras.Input(shape=(32, 32, 3)),

    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'),

    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),

    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(64, activation='relu'),

    tf.keras.layers.Dense(10, activation='softmax')

])

model.summary()
```

```python
# Compile the CNN model

model.compile(

    optimizer='adam',

    loss='sparse_categorical_crossentropy',

    metrics=['accuracy']

)

print("Model compiled successfully")
```

```python
# Train the CNN model

history = model.fit(

    X_train,

    y_train,

    epochs=10,

    validation_data=(X_test, y_test)

)
```

```python
# Evaluate the CNN model

test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", test_loss)

print("Test Accuracy:", test_accuracy)
```

```python
# Make predictions on test images

predictions = model.predict(X_test)

predicted_classes = predictions.argmax(axis=1)

print("First 10 Predicted Classes:")

print(predicted_classes[:10])

print("First 10 Actual Classes:")

print(y_test[:10].flatten())
```

```python
# Display test images with predicted labels

import matplotlib.pyplot as plt

class_names = [

    'airplane', 'automobile', 'bird', 'cat', 'deer',

    'dog', 'frog', 'horse', 'ship', 'truck'

]

plt.figure(figsize=(10, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(X_test[i])

    plt.xticks([])

    plt.yticks([])

    plt.xlabel(class_names[predicted_classes[i]])

plt.tight_layout()

plt.show()
```

```python
# Plot training and validation accuracy

plt.plot(history.history['accuracy'], label='Training Accuracy')

plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.xlabel('Epoch')

plt.ylabel('Accuracy')

plt.title('CNN Training and Validation Accuracy')

plt.legend()

plt.show()
```

```python
# Plot training and validation loss

plt.plot(history.history['loss'], label='Training Loss')

plt.plot(history.history['val_loss'], label='Validation Loss')

plt.xlabel('Epoch')

plt.ylabel('Loss')

plt.title('CNN Training and Validation Loss')

plt.legend()

plt.show()
```

```python
# Final CNN results

print("===== CIFAR-10 CNN FINAL RESULTS =====")

print(f"Test Accuracy: {test_accuracy * 100:.2f}%")

print(f"Test Loss: {test_loss:.4f}")

print("CNN model training and evaluation completed successfully!")
```
