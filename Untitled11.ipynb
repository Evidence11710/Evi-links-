```python
import pandas as pd
import numpy as np
import re
import string

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import matplotlib.pyplot as plt

print("Step 1 completed successfully")
```

```python
import pandas as pd

df = pd.read_csv("canada_fake_news_dataset.csv")

print("Step 3 completed successfully")
print("Dataset shape:", df.shape)
print(df.head())
```

```python
print("Column names:")
print(df.columns)

print("\nMissing values:")
print(df.isnull().sum())

print("\nLabel counts:")
print(df["label"].value_counts())

df = df.dropna()

print("\nStep 4 completed successfully")
print("Dataset shape after cleaning:", df.shape)
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'\d+', ' ', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df["clean_text"] = df["text"].apply(clean_text)

print("Step 5 completed successfully")
print(df[["text", "clean_text"]].head())
```

```python
df["target"] = df["label"].map({
    "FAKE": 0,
    "REAL": 1
})

print("Step 6 completed successfully")
print(df[["label", "target"]].head())

print("\nTarget counts:")
print(df["target"].value_counts())
```

```python
X = df["clean_text"]
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Step 7 completed successfully")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_df=0.7,
    max_features=10000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Step 8 completed successfully")
print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)

print("Step 9 completed successfully")
```

```python
y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

print("Step 10 completed successfully")
print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
print("Classification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["FAKE", "REAL"]
    )
)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Step 12 completed successfully")
print("Confusion Matrix:")
print(cm)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(cm)
plt.title("Fake News Detection Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.xticks([0, 1], ["FAKE", "REAL"])
plt.yticks([0, 1], ["FAKE", "REAL"])

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar()
plt.show()

print("Step 13 completed successfully")
```

```python
def predict_news(news):
    cleaned_news = clean_text(news)
    vectorized_news = tfidf.transform([cleaned_news])

    prediction = model.predict(vectorized_news)[0]
    confidence = model.predict_proba(vectorized_news).max()

    if prediction == 1:
        result = "REAL NEWS"
    else:
        result = "FAKE NEWS"

    print("Prediction:", result)
    print("Confidence:", round(confidence * 100, 2), "%")
```

```python
sample_news = """
The Government of Canada announced new funding
for public infrastructure projects.
"""

predict_news(sample_news)
```
