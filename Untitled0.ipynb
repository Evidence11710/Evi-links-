```python
import numpy as np

import matplotlib.pyplot as plt

from tensorflow.keras.datasets import reuters

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Dense

from tensorflow.keras.utils import to_categorical

(train_data, train_labels), (test_data, test_labels) = reuters.load_data(num_words=10000)

print("Training Samples:", len(train_data))

print("Testing Samples:", len(test_data))
```

```python
print(train_data[0])
```

```python
print(train_labels[0])
```

```python
def vectorize_sequences(sequences, dimension=10000):

    results = np.zeros((len(sequences), dimension))

    for i, sequence in enumerate(sequences):

        results[i, sequence] = 1.0

    return results

x_train = vectorize_sequences(train_data)

x_test = vectorize_sequences(test_data)

print("Training data shape:", x_train.shape)

print("Testing data shape:", x_test.shape)
```

```python
one_hot_train_labels = to_categorical(train_labels)

one_hot_test_labels = to_categorical(test_labels)

print("Training labels shape:", one_hot_train_labels.shape)

print("Testing labels shape:", one_hot_test_labels.shape)
```

```python
model = Sequential()
```

```python
model.add(Dense(64, activation='relu', input_shape=(10000,)))
```

```python
model.add(Dense(64, activation='relu'))
```

```python
model.compile(

    optimizer='rmsprop',

    loss='categorical_crossentropy',

    metrics=['accuracy']

)
```

```python
model.summary()
```

```python
x_val = x_train[:1000]

partial_x_train = x_train[1000:]

y_val = one_hot_train_labels[:1000]

partial_y_train = one_hot_train_labels[1000:]
```

```python
history = model.fit(

    partial_x_train,

    partial_y_train,

    epochs=20,

    batch_size=512,

    validation_data=(x_val, y_val)

)
```

```python
results = model.evaluate(x_test, one_hot_test_labels)

print("Test Loss:", results[0])

print("Test Accuracy:", results[1])
```

```python
results = model.evaluate(x_test, one_hot_test_labels)

print("Test Loss:", results[0])

print("Test Accuracy:", results[1])
```

```python
predictions = model.predict(x_test)

print("Predicted Class:", np.argmax(predictions[0]))

print("Actual Class:", test_labels[0])
```

```python
plt.plot(history.history['accuracy'])

plt.plot(history.history['val_accuracy'])

plt.title("Model Accuracy")

plt.ylabel("Accuracy")

plt.xlabel("Epoch")

plt.legend(["Training", "Validation"])

plt.show()
```

```python
plt.plot(history.history['loss'])

plt.plot(history.history['val_loss'])

plt.title("Model Loss")

plt.ylabel("Loss")

plt.xlabel("Epoch")

plt.legend(["Training", "Validation"])

plt.show()
```

```python
model.save("Reuters_Multiclass_Model.h5")

print("Model saved successfully!")
```
