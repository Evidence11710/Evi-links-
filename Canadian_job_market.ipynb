```python
import pandas as pd

df = pd.read_csv(

    "/content/1410044401-eng 3.csv",

    skiprows=9,

    encoding="utf-8-sig"

)

df.head(10)
```

```python
df.columns = [

    "Occupation",

    "Statistic",

    "Q2_2025",

    "Q3_2025",

    "Q4_2025",

    "Q1_2026",

    "Q2_2026"

]

df.head(10)
```

```python
import numpy as np

# Convert only values that are actually numbers

for col in ["Q2_2025", "Q3_2025", "Q4_2025", "Q1_2026", "Q2_2026"]:

    df[col] = (

        df[col]

        .astype(str)

        .str.replace(",", "", regex=False)

        .str.replace(r"[A-Z]", "", regex=True)

    )



    df[col] = pd.to_numeric(df[col], errors="coerce")

df.head(10)
```

```python
# Keep only rows containing job vacancy information

vacancies = df[df["Statistic"] == "Job vacancies 6"].copy()

vacancies.head(10)
```

```python
# Fill occupation names down to the vacancy rows

df["Occupation"] = df["Occupation"].ffill()

# Select job vacancy rows

vacancies = df[df["Statistic"] == "Job vacancies 6"].copy()

vacancies.head(10)
```

```python
# Calculate the change in job vacancies from Q2 2025 to Q2 2026

vacancies["Change"] = vacancies["Q2_2026"] - vacancies["Q2_2025"]

vacancies[["Occupation", "Q2_2025", "Q2_2026", "Change"]].head(10)
```

```python
# Top 10 occupations with the largest increase in job vacancies

top_increases = vacancies.sort_values(

    by="Change",

    ascending=False

).head(10)

top_increases[["Occupation", "Q2_2025", "Q2_2026", "Change"]]
```

```python
# Top 10 occupations with the largest decrease in job vacancies

top_decreases = vacancies.sort_values(

    by="Change",

    ascending=True

).head(10)

top_decreases[["Occupation", "Q2_2025", "Q2_2026", "Change"]]
```

```python
import matplotlib.pyplot as plt

chart_data = top_decreases.head(5)

plt.figure(figsize=(10, 6))

plt.bar(

    chart_data["Occupation"],

    chart_data["Change"]

)

plt.title("Top 5 Occupations with the Largest Decrease in Job Vacancies")

plt.xlabel("Occupation")

plt.ylabel("Change in Job Vacancies")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

```python
# Calculate the total change in job vacancies

total_change = (

    vacancies["Q2_2026"].iloc[0]

    - vacancies["Q2_2025"].iloc[0]

)

percentage_change = (

    total_change / vacancies["Q2_2025"].iloc[0]

) * 100

print("Total change in job vacancies:", total_change)

print("Percentage change:", round(percentage_change, 2), "%")
```

```python
print("CAPSTONE PROJECT SUMMARY")

print("=" * 40)

print("Total job vacancies in Q2 2025:", vacancies["Q2_2025"].iloc[0])

print("Total job vacancies in Q2 2026:", vacancies["Q2_2026"].iloc[0])

print("Overall change:", vacancies["Change"].iloc[0])

print("\nLargest increase:")

print(

    top_increases.iloc[0]["Occupation"],

    ":",

    top_increases.iloc[0]["Change"]

)

print("\nLargest decrease:")

print(

    top_decreases.iloc[0]["Occupation"],

    ":",

    top_decreases.iloc[0]["Change"]

)
```
